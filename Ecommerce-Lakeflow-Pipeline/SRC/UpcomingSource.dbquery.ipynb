```sql
USE CATALOG ecomm;
USE SCHEMA src;

CREATE TABLE customers (
    customer_id INT,
    name STRING,
    email STRING,
    city STRING,
    updated_at TIMESTAMP
);

CREATE TABLE products (
    product_id STRING,
    product_name STRING,
    category STRING,
    price DECIMAL(10,2),
    updated_at TIMESTAMP
);

CREATE TABLE orders (
    order_id BIGINT,
    customer_id INT,
    product_id STRING,
    quantity INT,
    amount DECIMAL(10,2),
    order_date TIMESTAMP
);
```
